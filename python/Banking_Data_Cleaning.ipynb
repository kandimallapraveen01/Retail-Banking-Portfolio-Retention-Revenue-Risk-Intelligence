```python
import pandas as pd
df = pd.read_csv("churn.csv.csv")
print(df.head())
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df["estimated_salary"].isnull().sum())
```

```python
mean_balance = df["balance"].mean()
std_balance = df["balance"].std()
upper_limit = mean_balance + (5 * std_balance)
print("Upper Limit:", upper_limit)
```

```python
df = df[df["balance"] <= upper_limit]
print(df.shape)
print(df["balance"].max())
```

```python
df.to_csv("cleaned_bank_customers.csv", index=False)
print("Cleaned dataset saved successfully.")
```
